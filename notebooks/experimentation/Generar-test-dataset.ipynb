```python
#Importar librerias para manejo de datos

import pandas as pd
import numpy as np

#Seleccionar el numero de records
n = 10000
np.random.seed(42)

#Definir las columnnas del dataset
df = pd.DataFrame({
    "edad": np.random.randint(18, 70, n),
    "ingresos_declarados": np.random.normal(500, 200, n).clip(50),
    "nivel_endeudamiento": np.random.uniform(0, 1, n),
    "utilizacion_tarjetas": np.random.randint(0, 100, n),
    "score_buro": np.random.randint(0, 100, n),
    "morosidad_prev": np.random.randint(0, 120, n),
    "consumo_electrico": np.random.randint(30, 400, n),
    "pago_serv_publicos": np.random.choice([0,1], n, p=[0.2,0.8]),
    "titularidad_serv_publico": np.random.choice([0,1], n, p=[0.3,0.7]),
    "remesas": np.random.choice([0, 100, 200, 300], n),
    "ingresos_bancarios": np.random.normal(700, 300, n).clip(50),
    "historial_empresa": np.random.choice([0,1], n, p=[0.3,0.7]),
    "portabilidad": np.random.choice([0,1], n, p=[0.3,0.7]),
    "ultimo_consumo_movil": np.random.randint(0, 100, n),
})

#Crear la variable decision segun reglas de negocio
df["pago"] = (
    (df["score_buro"] > 40) &
    (df["nivel_endeudamiento"] < 0.6) &
    (df["pago_serv_publicos"] == 1)
).astype(int)

#Guardar dataset como csv
df.to_csv("g1tfm-test-dataset.csv", index=False)
df.head()
```
